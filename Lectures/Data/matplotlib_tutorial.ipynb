```python
# Real Data

# Oscar Winner Movies
oscar_movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232] # in $M
# plot
plt.plot(years, oscar_revenue)
plt.title("Oscar Movies Revenue in each Year")
plt.xlabel("Years")
plt.ylabel("Revenue (in $M)")
```

```python
# Non-Oscar Winner Movies
non_oscar_movies = ["Slumdog Millionare", "Avatar", "Inception", "Hugo", "Lincoln"]
non_oscar_revenue = [378, 2788, 829, 185, 275] # in $M

plt.plot(years, oscar_revenue, "o--r", label="Oscar Movies")
plt.plot(years, non_oscar_revenue, "o--b", label="Non-Oscar Movies")
plt.title("Oscar Movies vs Non-Oscar Movies Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")
plt.legend()

plt.savefig("final_plot.png")
```

```python
# Bar charts

import numpy as np

movies = [
    "The Dark Knight",
    "The Hurt Locker",
    "The King's Speech",
    "The Artist",
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
revenue = [1005, 170, 427, 133, 232] # in $M

width = 0.4
x = np.arange(len(years))

plt.bar(x-width/2, revenue, width, label="Oscar", color="r")
plt.bar(x+width/2, non_oscar_revenue, width, label="Non-Oscar", color="g")
plt.title("Oscar vs Non-Oscar Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue (in $M)")

#for i in range(len(years)):
#    plt.text(years[i], revenue[i]+10, str(revenue[i]), ha="center")

plt.ylim(0, max(max(revenue), max(non_oscar_revenue))+100)
```

```python
# Horizontal Chart
plt.barh(movies, revenue)
plt.xlabel("Revenue")
plt.ylabel("Movie Names")
```

```python
# Scatter Plot
people = ["Person A", "Person B", "Person C", "Person D", "Person E",
         "Person F", "Person G", "Person H", "Person I", "Person J"]
age = [22, 25, 30, 35, 40, 45, 50, 55, 60, 65]
bp = [110, 115, 120, 122, 125, 130, 135, 123, 145, 150]

colors = ["green" if x < 135 else "red" for x in bp]
plt.scatter(age, bp, s=bp, marker="o", alpha=1, cmap="OrRd", c=bp)
plt.title("Age vs BP")
plt.xlabel("Age")
plt.ylabel("Blood Pressure")
plt.grid()
plt.colorbar(label="BP")

for i in range(len(people)):
    plt.annotate(people[i], xy=(age[i], bp[i]), xytext=(age[i]+1, bp[i]+1))
    plt.xlim(min(age), max(age)+10)
    plt.ylim(min(bp), max(bp)+5)
```

```python
# Pie Charts
expenses = ["Salaries", "Rent", "Marketing", "R&D", "Miscellaneous"]
amounts = [500, 150, 120, 100, 50]

explode = [0, 0, 0, 0.15, 0]
plt.pie(amounts, labels=expenses, autopct="%1.1f%%", explode=explode, shadow=True)
```
