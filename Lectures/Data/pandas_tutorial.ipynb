```python
# AQI Data Set
df = pd.read_csv("globalAirQuality.csv")
df.head()
df.tail()
df.info()
df.describe()
```

```python
# Selecting Data

#columns
df["country"]  #view of the data
df[["city", "aqi"]]

#rows
df.loc[2]
df.loc[0:2]  #start idx : end idx (inclusive)
df.iloc[2]
df.iloc[0:2]  #start idx : end idx (exclusive)

#cells - rows & columns
df.loc[0:2, ["city", "aqi"]]
df.columns
df.iloc[0:3, 1:5]

#cell
df.at[0, "city"]
df.iat[0, 2]

cities = df["city"].copy()
cities[0] = "Mumbai"
cities
```

```python
# Filtering of Data
df[(df["aqi"] > 100) & (df["temperature"] > 30)]
df[(df["aqi"] > 100) & (df["temperature"] < 30)]
aqi_data = df[(df["aqi"] > 100) & (df["temperature"] > 30)][["city", "aqi"]]
aqi_data
aqi_data.loc[6]
aqi_data.iloc[0]
```

```python
# Query Method - returns copy
aqi_val = 100
aqi_data = df.query("aqi > @aqi_val & temperature > 30")[["city", "aqi"]].copy()
```

```python
# Cleaning Data - missing values
df = pd.read_csv("raw_data.csv")
df
df.isnull()
df.isna().sum()
df.dropna() #row drop
df.dropna(axis=1) #col drop
cleaned_data = df.copy()
cleaned_data["age"] = cleaned_data["age"].fillna(cleaned_data["age"].mean())
cleaned_data["income"] = cleaned_data["income"].fillna(cleaned_data["income"].mean())
cleaned_data["name"] = cleaned_data["name"].ffill()
cleaned_data["country"] = cleaned_data["country"].bfill()
cleaned_data["gender"] = cleaned_data["gender"].ffill()
cleaned_data
```

```python
# Handle Duplicates
df[["country", "gender"]].duplicated()
df2 = df.copy()
df2.drop_duplicates(inplace=True)
df2
```

```python
# Data Types
df.dtypes
df2 = df.copy()
df2 = df2.fillna(0)
df2["age"] = df2["age"].astype("int64")
df2.dtypes
```
